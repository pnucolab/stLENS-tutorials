```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import warnings

stlens_sec = {
    '10K':  [2*60+50,        3*60+6,         2*60+42],
    '30K':  [7*60+2,         6*60+48,        7*60+3],
    '50K':  [11*60+40,       11*60+9,        11*60+47],
    '100K': [15*60+32,       16*60+8,        15*60+53],
    '500K': [1*3600+16*60+29, 1*3600+20*60+55, 1*3600+29*60+36],
    '1M':   [3*3600+37*60+58, 3*3600+23*60+31, 3*3600+50*60+29],
}

sclens_sec = {
    '10K':  [298.365795,  301.701874,  565.511306],
    '30K':  [1652.496252, 1507.571076, 1395.340913],
    '50K':  [2337.012090, 1891.501286, 1557.966448],
    '100K': [2504.151383, 2323.581308, 2417.729743],
    '500K': [np.nan,      np.nan,      np.nan],
    '1M':   [np.nan,      np.nan,      np.nan],
}

data_sizes = ['10K', '30K', '50K', '100K', '500K', '1M']
x = np.arange(len(data_sizes))

def mean_std_hours(d):
    arr = np.array([d[s] for s in data_sizes], dtype=float) / 3600.0
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        return np.nanmean(arr, axis=1), np.nanstd(arr, axis=1, ddof=1)

stlens_mean, stlens_std = mean_std_hours(stlens_sec)
sclens_mean, sclens_std = mean_std_hours(sclens_sec)

colors = {'stLENS': '#77b5d9', 'scLENS': '#cc82f4'}

plt.figure(figsize=(6, 8))

plt.errorbar(
    x, stlens_mean, yerr=stlens_std,
    fmt='o-', color=colors['stLENS'], ecolor=colors['stLENS'],
    elinewidth=0.8, markeredgewidth=0.8, capsize=8,
)
plt.errorbar(
    x, sclens_mean, yerr=sclens_std,
    fmt='o-', color=colors['scLENS'], ecolor=colors['scLENS'],
    elinewidth=0.8, markeredgewidth=0.8, capsize=8,
)

legend_handles = [
    Line2D([0], [0], color=colors['stLENS'], marker='o', label='stLENS'),
    Line2D([0], [0], color=colors['scLENS'], marker='o', label='scLENS'),
]

plt.xticks(x, data_sizes)
plt.xlabel("Data Size (Number of Cells)")
plt.ylabel("Execution Time (h)")
plt.title("Execution Time by Data Size")
plt.legend(handles=legend_handles, loc='upper left', frameon=False)
plt.grid(False)
plt.tight_layout()
plt.savefig('/bce/groups/pnucolab/analysis/stLENS/dataset/paper/execute/sc_st_time.pdf', dpi=300)
plt.show()
```
