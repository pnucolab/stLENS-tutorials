```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
import warnings

stlens_sec = {
    '10K':  [2*60+50,        3*60+6,         2*60+42],
    '30K':  [7*60+2,         6*60+48,        7*60+3],
    '50K':  [11*60+40,       11*60+9,        11*60+47],
    '100K': [15*60+32,       16*60+8,        15*60+53],
    '500K': [1*3600+16*60+29, 1*3600+20*60+55, 1*3600+29*60+36],
    '1M':   [3*3600+37*60+58, 3*3600+23*60+31, 3*3600+50*60+29],
}

sclens_sec = {
    '10K':  [298.365795,  301.701874,  565.511306],
    '30K':  [1652.496252, 1507.571076, 1395.340913],
    '50K':  [2337.012090, 1891.501286, 1557.966448],
    '100K': [2504.151383, 2323.581308, 2417.729743],
    '500K': [np.nan,      np.nan,      np.nan],
    '1M':   [np.nan,      np.nan,      np.nan],
}

data_sizes = ['10K', '30K', '50K', '100K', '500K', '1M']
x = np.arange(len(data_sizes))

def mean_std_hours(d):
    arr = np.array([d[s] for s in data_sizes], dtype=float) / 3600.0
    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        return np.nanmean(arr, axis=1), np.nanstd(arr, axis=1, ddof=1)

stlens_mean, stlens_std = mean_std_hours(stlens_sec)
sclens_mean, sclens_std = mean_std_hours(sclens_sec)

colors = {'stLENS': '#77b5d9', 'scLENS': '#cc82f4'}

plt.figure(figsize=(6, 8))

plt.errorbar(
    x, stlens_mean, yerr=stlens_std,
    fmt='o-', color=colors['stLENS'], ecolor=colors['stLENS'],
    elinewidth=0.8, markeredgewidth=0.8, capsize=8,
)
plt.errorbar(
    x, sclens_mean, yerr=sclens_std,
    fmt='o-', color=colors['scLENS'], ecolor=colors['scLENS'],
    elinewidth=0.8, markeredgewidth=0.8, capsize=8,
)

legend_handles = [
    Line2D([0], [0], color=colors['stLENS'], marker='o', label='stLENS'),
    Line2D([0], [0], color=colors['scLENS'], marker='o', label='scLENS'),
]

plt.xticks(x, data_sizes)
plt.xlabel("Data Size (Number of Cells)")
plt.ylabel("Execution Time (h)")
plt.title("Execution Time by Data Size")
plt.legend(handles=legend_handles, loc='upper left', frameon=False)
plt.grid(False)
plt.tight_layout()
plt.savefig('./stLENS/dataset/paper/execute/sc_st_time.pdf', dpi=300)
plt.show()
```

# Memory Graph

```python
import json, os
from datetime import datetime, timezone, timedelta
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

KST = timezone(timedelta(hours=9))

def _parse_kst(t):
    """문자열/datetime → tz-aware KST datetime. None은 그대로 None."""
    if t is None:
        return None
    if isinstance(t, datetime):
        return t.replace(tzinfo=KST) if t.tzinfo is None else t.astimezone(KST)
    for fmt in ('%Y-%m-%d %H:%M:%S', '%Y-%m-%d %H:%M', '%Y-%m-%d'):
        try:
            return datetime.strptime(t, fmt).replace(tzinfo=KST)
        except ValueError:
            continue
    raise ValueError(f"시간 문자열 파싱 실패: {t}")

def plot_memory_all(entries, save_dir='.'):
    color_map = {
        '10k': '#2ca02c', '10K': '#2ca02c',
        '50k': '#1f77b4', '50K': '#1f77b4',
        '100k': '#ff7f0e', '100K': '#ff7f0e',
    }

    fig, ax = plt.subplots(figsize=(15.6, 3.8), dpi=100)
    max_elapsed = 0

    for json_path, start, end, label in entries:
        with open(json_path) as f:
            data = json.load(f)['data']

        records = [(datetime.fromtimestamp(e['time'], tz=KST), e.get('mem', 0)) for e in data]
        records.sort(key=lambda x: x[0])

        t_start = _parse_kst(start)
        t_end   = _parse_kst(end)
        records = [(t, m) for t, m in records if t_start <= t <= t_end]

        times  = [t for t, _ in records]
        mem_gb = np.array([m for _, m in records], dtype=float) / (1024 ** 3)

        real_mask = mem_gb > 0
        if not real_mask.any():
            continue

        first_nonzero = int(np.argmax(real_mask))
        proc_start_idx = max(0, first_nonzero - 1)

        times  = times[proc_start_idx:]
        mem_gb = mem_gb[proc_start_idx:]
        proc_start_time = times[0]
        elapsed_min = np.array([(t - proc_start_time).total_seconds() / 60 for t in times])

        max_elapsed = max(max_elapsed, elapsed_min[-1])

        color = color_map.get(label, '#333333')
        ax.plot(elapsed_min, mem_gb, color=color, lw=1.8, label=label)
        ax.fill_between(elapsed_min, mem_gb, alpha=0.1, color=color)

        peak_idx = int(np.argmax(mem_gb))
        ax.plot(elapsed_min[peak_idx], mem_gb[peak_idx],
                'o', color=color, markersize=6, zorder=5)
        ax.annotate(
            f'{mem_gb[peak_idx]:.2f} GB',
            xy=(elapsed_min[peak_idx], mem_gb[peak_idx]),
            xytext=(5, 7), textcoords='offset points',
            fontsize=11, color=color,
            fontweight='bold',
        )

    ax.set_xlabel('Elapsed time (min)')
    ax.set_ylabel('CPU Memory (GB)')
    ax.set_xlim(0, max_elapsed)
    ax.set_ylim(0, 40)
    ax.legend(loc='upper right', frameon=True)
    ax.grid(linestyle='--', lw=0.3)
    fig.tight_layout()

    fname = os.path.join(save_dir)
    fig.savefig(fname, dpi=150, bbox_inches='tight')
    print(f'Saved: {fname}')
    return fig
```

```python
entries_gpu = [
    ('./stLENS/dataset/paper/execute/memory/prox_10k_gpu_shifted.json',
     '2026-04-22 21:35', '2026-04-22 21:40', '10k'),
    ('./stLENS/dataset/paper/execute/memory/prox_50k_gpu.json',
     '2026-04-24 2:56',  '2026-04-24 3:20',  '50k'),
    ('./stLENS/dataset/paper/execute/memory/prox_0424.json',
     '2026-04-24 11:47', '2026-04-24 12:15', '100k'),
]

plot_memory_all(entries_gpu, save_dir='./stLENS/dataset/paper/execute/memory/gpu_memory.pdf')
```

```python
entries_cpu= [
    ('./stLENS/dataset/paper/execute/memory/prox_10k_cpu.json',
     '2026-04-24 2:25', '2026-04-24 3:15', '10k'),
    ('./stLENS/dataset/paper/execute/memory/prox_0424_2.json',
     '2026-04-24 12:17',  '2026-04-24 13:17',  '50k'),
    ('./stLENS/dataset/paper/execute/memory/prox_0424.json',
     '2026-04-24 3:20', '2026-04-24 5:02', '100k'),
]
plot_memory_all(entries_cpu, save_dir='./stLENS/dataset/paper/execute/memory/cpu_memory.pdf')
```
