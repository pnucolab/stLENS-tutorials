# Jackstraw

```python
import numpy as np
import pandas as pd
import scanpy as sc
from scipy.sparse import issparse

import rpy2.robjects as ro
from rpy2.robjects import numpy2ri
from rpy2.robjects.conversion import localconverter
from rpy2.robjects.packages import importr

import warnings
warnings.filterwarnings('ignore')
print("scanpy:", sc.__version__)
Seurat = importr('Seurat')
print("Seurat OK")
```

## stomics


```python
adata_js = sc.read_h5ad('/bce/groups/pnucolab/analysis/scLENS/scLENS_paper_yh/data/stomics/E16.5_E1S3_cell_bin_whole_brain.h5ad')
print(adata_js)
```

```python
adata_js.X = adata_js.layers["counts"].copy()
sc.pp.filter_cells(adata_js, min_genes=200)
sc.pp.filter_genes(adata_js, min_cells=15)
print("after filter:", adata_js.shape)
```

```python
counts = adata_js.layers['counts']
counts_np = counts.toarray() if issparse(counts) else np.asarray(counts)
counts_gc = counts_np.T.astype(np.float64)
print(f'gene x cell shape: {counts_gc.shape}, mem: {counts_gc.nbytes/1024**3:.2f} GB')
gene_names = list(adata_js.var_names)
cell_names = list(adata_js.obs_names)
```

```python
with localconverter(ro.default_converter + numpy2ri.converter):
    ro.globalenv['counts_r'] = ro.conversion.py2rpy(counts_gc)
ro.globalenv['gene_names'] = ro.StrVector(gene_names)
ro.globalenv['cell_names'] = ro.StrVector(cell_names)
```

```python
ro.r("""
rownames(counts_r) <- gene_names
colnames(counts_r) <- cell_names

library(Seurat)
library(future)
options(future.globals.maxSize = +Inf)  # 큰 매트릭스 워커 전송 한도 풀기
set.seed(42)

cat("CreateSeuratObject...\n")
so <- CreateSeuratObject(counts = counts_r, project = "jackstraw_brain")
cat(sprintf(" created: %d features x %d cells\n", nrow(so), ncol(so)))

cat("NormalizeData + FindVariableFeatures + ScaleData...\n")
so <- NormalizeData(so, verbose = FALSE)
so <- FindVariableFeatures(so, nfeatures = 2000, verbose = FALSE)
so <- ScaleData(so, verbose = FALSE)
cat("preprocessing done\n")
""")
```

```python
%%time
ro.r("""
so <- RunPCA(so, npcs = 100, verbose = FALSE)
""")
```

```python
%%time
ro.r("""
so <- JackStraw(so, num.replicate = 10, dims = 100, prop.freq = 0.05, verbose = FALSE)
""")
```

```python
%%time
ro.r("""
so <- ScoreJackStraw(so, dims = 1:100)
""")
```

```python
ro.r('rm(so, counts_r, gene_names, cell_names); gc()')
del adata_js, counts, counts_np, counts_gc
import gc; gc.collect()
```
