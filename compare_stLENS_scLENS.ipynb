# Violin plot: PC count comparison (stLENS vs scLENS)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

stlens_pc = [257, 257, 256, 259, 258, 259, 258, 259, 256, 256] 
sclens_pc = [256, 259, 258, 257, 259, 257, 259,258, 257, 256]


palette = {
    'stLENS': '#77b5d9',
    'scLENS': "#cc82f4",
}

df = pd.DataFrame({
    'method': ['stLENS'] * len(stlens_pc) + ['scLENS'] * len(sclens_pc),
    'n_pc'  : stlens_pc + sclens_pc,
})

plt.figure(figsize=(4, 5))
ax = sns.violinplot(
    data=df,
    x='method', y='n_pc',
    hue='method',
    palette=palette,
    legend=False,
    inner='box',
    linewidth=1.2,
    cut=0
)

ax.set_xlabel('')
ax.set_ylabel('Number of significant PCs')
sns.despine()
plt.tight_layout()

plt.savefig('./pc_count_violin.pdf', dpi=300, bbox_inches='tight')

plt.show()
```
